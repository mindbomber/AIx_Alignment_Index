# AIx Open Example
Load, validate, score, compare, and inspect a complete assessment.

```python
from pathlib import Path
from aix.io import load_document
from aix.models import validate_assessment
from aix.scoring import score_assessment, compare_results

root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
assessment = load_document(root / 'examples' / 'ai_output_assessment.yaml')
validate_assessment(assessment)
result = score_assessment(assessment)
result.to_dict()
```

```python
platform = score_assessment(load_document(root / 'examples' / 'platform_assessment.yaml'))
compare_results(result, platform)
```
